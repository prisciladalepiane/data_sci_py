```python
from dataclasses import dataclass
```


Dataclass é um recurso do Python usado para criar classes cujo principal objetivo é armazenar dados. Ela reduz código repetitivo, porque gera automaticamente métodos como `__init__`, `__repr__` e `__eq__`.

O `__init__` é o método responsável por inicializar o objeto quando ele é criado. É nele que normalmente são definidos os atributos da classe.

O `__repr__` define como o objeto será representado em texto. Ele serve para mostrar uma descrição mais legível do objeto, em vez de aparecer apenas uma referência de memória.

O `__eq__` define como dois objetos serão comparados usando ==. Ele determina quais atributos precisam ser iguais para que dois objetos sejam considerados iguais.

`@dataclass` é usada para reduzir código repetitivo em **classes que servem principalmente para armazenar dados**. Ela gera automaticamente métodos como construtor, representação do objeto e comparação entre objetos.

> É uma maneira mais prática de criar classes que funcionam como estruturas de dados.

Exemplo, criando a Classe Produto sem o dataclass:

```python
class Produto:
    def __init__(self, nome: str, preco: float, estoque: int):
        self.nome = nome
        self.preco = preco
        self.estoque = estoque

    def __repr__(self):
        return (
            f"Produto(nome={self.nome!r}, "
            f"preco={self.preco!r}, "
            f"estoque={self.estoque!r})"
        )

    def __eq__(self, outro):
        if not isinstance(outro, Produto):
            return False

        return (
            self.nome == outro.nome
            and self.preco == outro.preco
            and self.estoque == outro.estoque
        )
```

```python
produto1 = Produto("Mouse", 150.0, 10)
produto2 = Produto("Mouse", 150.0, 10)

print(produto1)
print(produto1 == produto2)
```

com dataclasses:

```python
@dataclass
class Produto:
    nome: str
    preco: float
    estoque: int
```

```python
produto1 = Produto("Teclado", 150.0, 10)
produto2 = Produto("Teclado", 150.0, 10)

print(produto1)
print(produto1 == produto2)
```

Usar opções como `order=True` cria métodos de comparação de ordem, como menor que, maior que, menor ou igual e maior ou igual.


```python
@dataclass(order=True)
class Produto:
    preco: float
    nome: str
```

```python
p1 = Produto(100.0, "Mouse")
p2 = Produto(200.0, "Teclado")

print(p1 < p2)
```


Com `frozen=True`, ela impede que os atributos sejam alterados depois que o objeto é criado.

```python
@dataclass(frozen=True)
class Produto:
    nome: str
    preco: float
```

```python
produto = Produto("Teclado", 150.0)

produto.preco = 200.0
```

Isso gera <strong style="color: red;">erro</strong> pois o objeto foi definido como imutável.

Dependendo da configuração, ela também pode gerar um `__hash__`, que permite que objetos sejam usados em estruturas como conjuntos (set) ou como chaves de dicionários.

```python
@dataclass(frozen=True)
class Produto:
    nome: str
    preco: float

produto1 = Produto("Teclado", 150.0)

produto1 = {produto1}
```

```python
produto = Produto("Teclado", 150.0)
produto2 = Produto("Telefone", 1250.0)

estoque = {
    produto: 1,
    produto2: 2
}

print(estoque)
```
